## Missing Values

```python
# Checking missing values
df.isnull().sum()
```

```python
# handling Missing Values in CouponCode column
df["CouponCode"] = df["CouponCode"].fillna("NO COUPON")
df["CouponCode"] = df["CouponCode"].str.upper()
```

### Unique Value validation
Unique values within categorical columns were reviewed to identify inconsistency in text formatting and spelling.

```python
# Checking unique values to identify inconsistencies in text_columns
text_columns = ["Product", "PaymentMethod", "OrderStatus", "ReferralSource", "ShippingAddress"]

for col in text_columns:
    print(f"\nUnique values in {col}:")
    print (df[col].unique())
```

### Text Standardization
Text columns were cleaned by removing extra spaces and standardizing text formatting for consistency.

```python
# Removing extra spaces and standardizing text_columns
for col in text_columns:
    df[col] = df[col].str.strip()
    df[col] = df[col].str.capitalize()
```

##  Validation of Numerical Calculation
A validation check was performed on the TotalPrice column to ensure that revenue values matched the expected calculation based on Quantity and UnitPrice

```python
# Validating TotalPrice by creating a check column
df["CalculatedTotal"] = df["Quantity"] * df["UnitPrice"]
```

```python
mismatch = df[df["CalculatedTotal"] != df["TotalPrice"]] [["OrderID", "Quantity", "UnitPrice", "TotalPrice", "CalculatedTotal"]]
print(mismatch)
```

```python
df.drop(columns = "CalculatedTotal", inplace = True)
```
